```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
```

```python
#3D Surface Plot
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111,  projection='3d')

x = np.linspace(-5, 5, 60)
y = np.linspace(-5, 5, 60)
X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

surf = ax.plot_surface(X, Y, Z,
        cmap='viridis',
        edgecolor='none',
        alpha=0.85)
fig.colorbar(surf, ax=ax, shrink=0.85, label='Z value')
ax.set_title('3D Surface: sin(√(x²+y²))', fontsize=13)
ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')
ax.view_init(elev=30, azim=120) #camera angle
plt.savefig('3D Surface: sin(√(x²+y²))', bbox_inches='tight', facecolor='white', dpi=150)
plt.show()
```

```python
#3D Scatter
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
n = 200
ax.scatter(np.random.randn(n), np.random.randn(n), np.random.randn(n), c=np.random.rand(n), cmap='plasma', s = 30, alpha = 0.7)
ax.set_title('3D Scatter Plot')
plt.show()
```
